```python
class Database:
  def __init__(self):
    self.data_base = {}
  def add_user(self, username, password):
    self.data_base[username] = password

class User:
  ''' Класс User принимает атрибуты Логин, Пароль '''
  def __init__(self, username, password, confirm_password):
    self.username = username
    if password == confirm_password:
      self.password = password

if __name__ == '__main__':
  data_base = Database()
  user = User(input('Введите логин: '), input('Введите пароль: '), input('Повторите пароль: '))
  data_base.add_user(user.username, user.password)
```
